```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the CSV file
df = pd.read_csv('profiler_logs/timing_data_fastloader.csv')

# Set figure size and style
plt.figure(figsize=(12, 8))
sns.set_style("whitegrid")

# Create box plots for the three metrics
plt.subplot(1, 3, 1)
sns.boxplot(y=df['Data_Wait_Time_ms'], color='skyblue')
plt.title('Data Wait Time (ms)')
plt.ylabel('Time (ms)')
plt.grid(True, linestyle='--', alpha=0.7)

plt.subplot(1, 3, 2)
sns.boxplot(y=df['Compute_Time_ms'], color='lightgreen')
plt.title('Compute Time (ms)')
plt.ylabel('Time (ms)')
plt.grid(True, linestyle='--', alpha=0.7)

plt.subplot(1, 3, 3)
sns.boxplot(y=df['Stall_Percentage'], color='salmon')
plt.title('Stall Percentage (%)')
plt.ylabel('Percentage (%)')
plt.grid(True, linestyle='--', alpha=0.7)

# Adjust layout and add a title
plt.tight_layout()
plt.suptitle('Distribution of Timing Metrics', fontsize=16, y=1.05)

# Show the plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Read both CSV files
fast_loader_df = pd.read_csv('/home/cc/fastLoader/profiler_logs/timing_data.csv')
pytorch_df = pd.read_csv('/home/cc/fastLoader/pytorch_profiler_logs/timing_data.csv')

# Print basic statistics
print("FastLoader Statistics:")
print(fast_loader_df[['Data_Wait_Time_ms', 'Compute_Time_ms', 'Stall_Percentage']].describe())

print("\nPyTorch Statistics:")
print(pytorch_df[['Data_Wait_Time_ms', 'Compute_Time_ms', 'Stall_Percentage']].describe())

# Create figure for comparison
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Data Wait Time comparison
ax = axes[0]
box_data = [fast_loader_df['Data_Wait_Time_ms'], pytorch_df['Data_Wait_Time_ms']]
ax.boxplot(box_data, labels=['FastLoader', 'PyTorch'])
ax.set_title('Data Wait Time (ms)')
ax.set_ylabel('Time (ms)')
ax.grid(True, linestyle='--', alpha=0.7)

# Compute Time comparison
ax = axes[1]
box_data = [fast_loader_df['Compute_Time_ms'], pytorch_df['Compute_Time_ms']]
ax.boxplot(box_data, labels=['FastLoader', 'PyTorch'])
ax.set_title('Compute Time (ms)')
ax.set_ylabel('Time (ms)')
ax.grid(True, linestyle='--', alpha=0.7)

# Stall Percentage comparison
ax = axes[2]
box_data = [fast_loader_df['Stall_Percentage'], pytorch_df['Stall_Percentage']]
ax.boxplot(box_data, labels=['FastLoader', 'PyTorch'])
ax.set_title('Stall Percentage (%)')
ax.set_ylabel('Percentage (%)')
ax.grid(True, linestyle='--', alpha=0.7)

plt.suptitle('FastLoader vs PyTorch Timing Comparison', fontsize=16)
plt.tight_layout()
plt.show()

# Create histograms for a different view
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Data Wait Time histogram
ax = axes[0]
ax.hist(fast_loader_df['Data_Wait_Time_ms'], alpha=0.5, label='FastLoader')
ax.hist(pytorch_df['Data_Wait_Time_ms'], alpha=0.5, label='PyTorch')
ax.set_title('Data Wait Time Distribution')
ax.set_xlabel('Time (ms)')
ax.set_ylabel('Frequency')
ax.legend()

# Compute Time histogram
ax = axes[1]
ax.hist(fast_loader_df['Compute_Time_ms'], alpha=0.5, label='FastLoader')
ax.hist(pytorch_df['Compute_Time_ms'], alpha=0.5, label='PyTorch')
ax.set_title('Compute Time Distribution')
ax.set_xlabel('Time (ms)')
ax.set_ylabel('Frequency')
ax.legend()

# Stall Percentage histogram
ax = axes[2]
ax.hist(fast_loader_df['Stall_Percentage'], alpha=0.5, label='FastLoader')
ax.hist(pytorch_df['Stall_Percentage'], alpha=0.5, label='PyTorch')
ax.set_title('Stall Percentage Distribution')
ax.set_xlabel('Percentage (%)')
ax.set_ylabel('Frequency')
ax.legend()

plt.suptitle('FastLoader vs PyTorch Distribution Comparison', fontsize=16)
plt.tight_layout()
plt.show()
```
